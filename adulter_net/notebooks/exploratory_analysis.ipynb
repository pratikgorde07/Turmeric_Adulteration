# AdulterNet — Exploratory Data Analysis


```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from config import Config

config = Config()
print('Config loaded. Device:', config.device)
```

```python
from src.data.loader import load_all_modalities, validate_sample_ids
dfs = load_all_modalities(config)
for name, df in dfs.items():
    print(f'{name}: {df.shape}')
```

```python
# Class distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, (name, df) in zip(axes, dfs.items()):
    df['Target'].value_counts().sort_index().plot(kind='bar', ax=ax)
    ax.set_title(f'{name} — Class Distribution')
    ax.set_xlabel('Adulteration Level (%)')
    ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# FTIR mean spectra per class
df_ftir = dfs['ftir']
feat_cols = [c for c in df_ftir.columns if c not in ['Sample_ID', 'Target']]
fig, ax = plt.subplots(figsize=(14, 5))
for level in config.class_labels:
    mask = df_ftir['Target'] == level
    mean_spec = df_ftir.loc[mask, feat_cols].mean()
    ax.plot(mean_spec.values, label=f'{level}%', alpha=0.8)
ax.set_title('Mean FTIR Spectra per Adulteration Level')
ax.set_xlabel('Wavenumber index')
ax.set_ylabel('Absorbance')
ax.legend()
plt.tight_layout()
plt.show()
```
